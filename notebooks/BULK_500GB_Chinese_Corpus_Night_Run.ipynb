# BULK 500GB Chinese Corpus Night Run

One-click launcher. Uses the `HF_TOKEN` Colab Secret, downloads large public corpus shards, uploads each shard to Google Drive, deletes local copies, and skips completed Drive shards on rerun.


```python
from google.colab import userdata
import os, requests
token = userdata.get('HF_TOKEN')
assert token, 'HF_TOKEN missing from Colab Secrets'
os.environ['HF_TOKEN'] = token
url = 'https://raw.githubusercontent.com/rayzh2012/Open-Source-Research-OSR-/master/tools/bulk_500gb_chinese_corpus_night_run.py'
r = requests.get(url, timeout=120)
r.raise_for_status()
print('Loaded bulk runner:', len(r.text), 'bytes')
exec(compile(r.text, url, 'exec'), globals())
```
